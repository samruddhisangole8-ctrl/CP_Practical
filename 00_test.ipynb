```python
marks = float(input('Enter your marks: '))

if 90 <= marks <= 100:
    grade = 'A'
elif 80 <= marks < 90:
    grade = 'B'
elif 60 <= marks < 80:
    grade = 'C'
elif 40 <= marks < 60:
    grade = 'D'
else:
    grade = 'F'

print('Your grade is: ' + grade)
```
